# 전처리

`df1`은 숙소 정보이므로 이번 MVP에서는 사용하지 않고, 관광지/음식점 데이터인 `df2`만 사용합니다. `spot_id`, `content_id`는 추후 API와 연결할 수 있어 유지합니다.


```python
import numpy as np
import pandas as pd
from pathlib import Path

TOUR_SPOT_FILENAME = "_TourSpot__202607062337.csv"
PATH_CANDIDATES = [
    Path(TOUR_SPOT_FILENAME),
    Path("modeling") / TOUR_SPOT_FILENAME,
    Path("../modeling") / TOUR_SPOT_FILENAME,
]

TOUR_SPOT_PATH = next((p for p in PATH_CANDIDATES if p.exists()), None)
if TOUR_SPOT_PATH is None:
    raise FileNotFoundError(f"{TOUR_SPOT_FILENAME} ??? ?? ? ????. ?? ?? ??: {Path.cwd()}")

DATA_DIR = TOUR_SPOT_PATH.parent
spots_raw = pd.read_csv(TOUR_SPOT_PATH)
spots_raw.head()
```

```python
TYPE_NAME = {
    12: "관광지",
    14: "문화시설",
    15: "축제공연행사",
    25: "여행코스",
    28: "레포츠",
    32: "숙박",
    38: "쇼핑",
    39: "음식점",
}

VISIT_TYPE_IDS = [12, 14, 15, 28, 38]
RESTAURANT_TYPE_ID = 39
EXCLUDED_TYPE_IDS = [25, 32]

spots = spots_raw.copy()
spots["type_name"] = spots["content_type_id"].map(TYPE_NAME).fillna("??")

spots = spots[
    spots["map_x"].notna()
    & spots["map_y"].notna()
    & (spots["map_x"] != 0)
    & (spots["map_y"] != 0)
].copy()

spots["overview"] = spots["overview"].fillna("").astype(str)
spots["name"] = spots["name"].fillna("").astype(str)
spots["address"] = spots["address"].fillna("").astype(str)
spots["img"] = spots["img"].fillna("").astype(str)

# ??(32)? ?? ??? ????(25)? ?? ?? ???? ?????.
spots = spots[~spots["content_type_id"].isin(EXCLUDED_TYPE_IDS)].copy()

# overview ???? ????? ???? ??? ??? ?????.
spots["embedding_text"] = (
    "장소명: " + spots["name"] + "\n"
    + "분류: " + spots["type_name"] + "\n"
    + "설명: " + spots["overview"]
)

spots[["spot_id", "content_id", "name", "type_name", "address", "map_x", "map_y", "overview"]].head()
```

# 임베딩

한국어 우선 MVP로 `jhgan/ko-sroberta-multitask`를 먼저 사용합니다. 모델을 사용할 수 없는 환경에서는 TF-IDF로 자동 대체하여 테스트를 진행합니다.


```python
import os
import pickle
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer

EMBEDDING_MODEL_NAME = "jhgan/ko-sroberta-multitask"
EMBEDDING_CACHE_PATH = DATA_DIR / "spot_embeddings_ko.pkl"

embedding_backend = None
embedding_model = None
spot_vectors = None
vectorizer = None

try:
    from sentence_transformers import SentenceTransformer

    embedding_backend = "sentence-transformers"
    embedding_model = SentenceTransformer(EMBEDDING_MODEL_NAME)

    if EMBEDDING_CACHE_PATH.exists():
        with open(EMBEDDING_CACHE_PATH, "rb") as f:
            cache = pickle.load(f)
        if cache.get("model_name") == EMBEDDING_MODEL_NAME and cache.get("spot_ids") == spots["spot_id"].tolist():
            spot_vectors = cache["vectors"]

    if spot_vectors is None:
        spot_vectors = embedding_model.encode(
            spots["embedding_text"].tolist(),
            normalize_embeddings=True,
            show_progress_bar=True,
        )
        with open(EMBEDDING_CACHE_PATH, "wb") as f:
            pickle.dump(
                {
                    "model_name": EMBEDDING_MODEL_NAME,
                    "spot_ids": spots["spot_id"].tolist(),
                    "vectors": spot_vectors,
                },
                f,
            )
except Exception as e:
    embedding_backend = "tfidf-fallback"
    print("SentenceTransformer is not available. Using TF-IDF fallback.")
    print("reason:", repr(e))
    vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))
    spot_vectors = vectorizer.fit_transform(spots["embedding_text"].tolist())

print("embedding backend:", embedding_backend)
print("vector shape:", getattr(spot_vectors, "shape", None))

```

```python
def embed_query(text):
    """??? ?? ??? ?? ??? ?? ??? ??? ?????."""
    if embedding_backend == "sentence-transformers":
        return embedding_model.encode([text], normalize_embeddings=True)
    return vectorizer.transform([text])


def get_similarity_scores(query_text):
    query_vector = embed_query(query_text)
    return cosine_similarity(query_vector, spot_vectors)[0]
```

# 추천 알고리즘

테마와 분위기는 임베딩 유사도로 고르고, 시간/이동수단/음식점 개수는 명확한 규칙으로 처리합니다. 현재 데이터에는 영업시간, 반려동물, 주차, 혼잡도 정보가 없으므로 이번 MVP에서는 사용하지 않습니다.


```python
from math import radians, sin, cos, sqrt, atan2
from datetime import datetime, timedelta

DURATION_CONFIG = {
    "2시간": {"minutes": 120, "restaurant_count": 0, "start_time": "10:00"},
    "반나절": {"minutes": 240, "restaurant_count": 1, "start_time": "10:00"},
    "하루": {"minutes": 480, "restaurant_count": 2, "start_time": "10:00"},
    "1박2일": {"minutes": 960, "restaurant_count": 4, "start_time": "10:00"},
}

TRANSPORT_CONFIG = {
    "도보": {"speed_kmh": 4, "extra_min": 0, "max_leg_km": 3.0},
    "자전거": {"speed_kmh": 12, "extra_min": 0, "max_leg_km": 8.0},
    "대중교통": {"speed_kmh": 18, "extra_min": 10, "max_leg_km": 20.0},
    "자차": {"speed_kmh": 35, "extra_min": 5, "max_leg_km": 50.0},
}

VISIT_MINUTES = {12: 60, 14: 60, 15: 90, 28: 90, 38: 45, 39: 60}

THEME_HINTS = {
    "역사": "신라 역사 문화유산 유적 사찰 왕릉 설화 전통",
    "야경": "야경 밤 조명 노을 월지 월정교 산책",
    "힐링": "자연 숲 휴식 조용한 산책 웰니스",
    "경치": "풍경 전망 바다 산 호수 사진 명소",
    "로컬": "황리단길 골목 시장 카페 지역 분위기 체험",
    "맛집": "음식 맛집 식당 카페 디저트 로컬 먹거리",
    "산책": "걷기 산책 길 거리 공원 둘레길",
    "문화": "문화 예술 박물관 전시 공연 체험",
}

COMPANION_HINTS = {
    "혼자": "혼자 천천히 둘러보기 좋은 조용한 여행",
    "연인": "연인과 데이트하기 좋은 분위기 좋은 여행",
    "가족": "가족과 함께 편하게 방문하기 좋은 여행",
    "친구": "친구와 함께 즐겁게 둘러보기 좋은 여행",
    "동료": "동료와 함께 부담 없이 방문하기 좋은 여행",
}


def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371.0
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = sin(dlat / 2) ** 2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2) ** 2
    return 2 * R * atan2(sqrt(a), sqrt(1 - a))


def travel_minutes(distance_km, transport):
    cfg = TRANSPORT_CONFIG[transport]
    return int(round((distance_km / cfg["speed_kmh"]) * 60 + cfg["extra_min"]))


def build_user_query(duration, companions, themes, transport):
    if isinstance(themes, str):
        themes = [themes]
    theme_text = ", ".join(themes)
    hint_text = " ".join(THEME_HINTS.get(theme, theme) for theme in themes)
    companion_text = COMPANION_HINTS.get(companions, companions)
    return (
        f"동행 유형은 {companions}이고, {duration} 동안 경주에서 {theme_text} 테마를 즐기는 여행. "
        f"{transport}로 이동 가능한 코스를 선호한다. "
        f"{companion_text}. {hint_text}"
    )


def add_similarity_column(query_text):
    scored = spots.copy()
    scored["similarity"] = get_similarity_scores(query_text)
    scored["visit_minutes"] = scored["content_type_id"].map(VISIT_MINUTES).fillna(60).astype(int)
    return scored




```

```python
def choose_next_spot(candidates, current_lat, current_lon, transport, used_ids):
    available = candidates[~candidates["spot_id"].isin(used_ids)].copy()
    if available.empty:
        return None

    available["distance_from_prev_km"] = available.apply(
        lambda row: haversine_km(current_lat, current_lon, row["map_y"], row["map_x"]),
        axis=1,
    )

    max_leg = TRANSPORT_CONFIG[transport]["max_leg_km"]
    near = available[available["distance_from_prev_km"] <= max_leg].copy()
    if near.empty:
        near = available.copy()

    near["route_score"] = near["similarity"] - (near["distance_from_prev_km"] * 0.015)
    return near.sort_values(["route_score", "similarity"], ascending=False).iloc[0]


def pick_restaurant(restaurants, current_lat, current_lon, transport, used_ids):
    available = restaurants[~restaurants["spot_id"].isin(used_ids)].copy()
    if available.empty:
        return None

    available["distance_from_prev_km"] = available.apply(
        lambda row: haversine_km(current_lat, current_lon, row["map_y"], row["map_x"]),
        axis=1,
    )
    max_leg = TRANSPORT_CONFIG[transport]["max_leg_km"]
    near = available[available["distance_from_prev_km"] <= max_leg].copy()
    if near.empty:
        near = available.copy()

    near["route_score"] = near["similarity"] - (near["distance_from_prev_km"] * 0.02)
    return near.sort_values(["route_score", "similarity"], ascending=False).iloc[0]


def restaurant_slots(duration, restaurant_count):
    if restaurant_count == 0:
        return []
    if duration == "반나절":
        return [2]
    if duration == "하루":
        return [2, 5]
    if duration == "1박2일":
        return [2, 5, 8, 11]
    return []

```

```python
def recommend_course(
    duration="하루",
    companions="연인",
    themes=None,
    transport="도보",
    start_lat=None,
    start_lon=None,
    top_n=40,
):
    """?? CSV ???? ??? ??? + ??? ??? ?????."""
    if themes is None:
        themes = ["역사"]
    if isinstance(themes, str):
        themes = [themes]

    if duration not in DURATION_CONFIG:
        raise ValueError(f"duration? {list(DURATION_CONFIG)} ? ???? ???.")
    if transport not in TRANSPORT_CONFIG:
        raise ValueError(f"transport? {list(TRANSPORT_CONFIG)} ? ???? ???.")

    config = DURATION_CONFIG[duration]
    total_budget = config["minutes"]
    restaurant_count = config["restaurant_count"]
    query_text = build_user_query(duration, companions, themes, transport)
    scored = add_similarity_column(query_text)

    visit_candidates = scored[scored["content_type_id"].isin(VISIT_TYPE_IDS)].nlargest(top_n, "similarity").copy()
    restaurant_candidates = scored[scored["content_type_id"] == RESTAURANT_TYPE_ID].nlargest(max(top_n, 30), "similarity").copy()

    if visit_candidates.empty:
        raise ValueError("?? ??? ??? ??? ????.")

    used_ids = set()
    route = []
    elapsed = 0

    if start_lat is None or start_lon is None:
        first = visit_candidates.iloc[0].copy()
        first["distance_from_prev_km"] = 0.0
        first["travel_minutes"] = 0
    else:
        first = choose_next_spot(visit_candidates, start_lat, start_lon, transport, used_ids)
        first["travel_minutes"] = travel_minutes(first["distance_from_prev_km"], transport)

    route.append(first)
    used_ids.add(int(first["spot_id"]))
    elapsed += int(first["travel_minutes"]) + int(first["visit_minutes"])
    current_lat, current_lon = float(first["map_y"]), float(first["map_x"])

    slots = restaurant_slots(duration, restaurant_count)
    inserted_restaurants = 0
    step = 1

    while elapsed < total_budget:
        should_insert_restaurant = inserted_restaurants < restaurant_count and step in slots

        if should_insert_restaurant:
            next_spot = pick_restaurant(restaurant_candidates, current_lat, current_lon, transport, used_ids)
            inserted_restaurants += 1
        else:
            next_spot = choose_next_spot(visit_candidates, current_lat, current_lon, transport, used_ids)

        if next_spot is None:
            break

        next_spot = next_spot.copy()
        if "distance_from_prev_km" not in next_spot or pd.isna(next_spot["distance_from_prev_km"]):
            next_spot["distance_from_prev_km"] = haversine_km(current_lat, current_lon, next_spot["map_y"], next_spot["map_x"])
        next_spot["travel_minutes"] = travel_minutes(float(next_spot["distance_from_prev_km"]), transport)

        added_time = int(next_spot["travel_minutes"]) + int(next_spot["visit_minutes"])
        if elapsed + added_time > total_budget:
            break

        route.append(next_spot)
        used_ids.add(int(next_spot["spot_id"]))
        elapsed += added_time
        current_lat, current_lon = float(next_spot["map_y"]), float(next_spot["map_x"])
        step += 1

    result = pd.DataFrame(route).reset_index(drop=True)
    result["order"] = result.index + 1

    start_dt = datetime.strptime(config["start_time"], "%H:%M")
    current_dt = start_dt
    arrivals, leaves, day_labels = [], [], []

    for _, row in result.iterrows():
        current_dt += timedelta(minutes=int(row["travel_minutes"]))
        arrivals.append(current_dt.strftime("%H:%M"))
        current_dt += timedelta(minutes=int(row["visit_minutes"]))
        leaves.append(current_dt.strftime("%H:%M"))
        day_labels.append(1 if int((current_dt - start_dt).total_seconds() // 60) <= 480 else 2)

    result["day"] = day_labels
    result["arrival_time"] = arrivals
    result["leave_time"] = leaves
    result["distance_from_prev_km"] = result["distance_from_prev_km"].astype(float).round(2)
    result["similarity"] = result["similarity"].astype(float).round(4)
    result["reason"] = result.apply(
        lambda row: f"'{', '.join(themes)}' ??? ??? ???? ??, ?? ???? ? {row['distance_from_prev_km']}km ??? ????.",
        axis=1,
    )

    columns = [
        "day", "order", "name", "type_name", "address", "similarity",
        "distance_from_prev_km", "travel_minutes", "visit_minutes",
        "arrival_time", "leave_time", "reason", "spot_id", "content_id", "content_type_id", "img",
    ]
    result = result[columns]

    summary = {
        "query": query_text,
        "duration": duration,
        "companions": companions,
        "themes": themes,
        "transport": transport,
        "total_places": len(result),
        "restaurant_count": int((result["content_type_id"] == RESTAURANT_TYPE_ID).sum()),
        "target_restaurant_count": restaurant_count,
        "total_distance_km": round(float(result["distance_from_prev_km"].sum()), 2),
        "total_time_minutes": int(result["travel_minutes"].sum() + result["visit_minutes"].sum()),
        "time_budget_minutes": total_budget,
        "embedding_backend": embedding_backend,
    }
    return result, summary

```
